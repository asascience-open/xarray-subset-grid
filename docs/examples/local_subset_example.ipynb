# Local Subset Example

Demonstrates subsetting using a bundled local NetCDF file.
No network access required.

```python
import xarray as xr
import numpy as np
from pathlib import Path
import xarray_subset_grid  # noqa: F401

data_path = Path('../../tests/example_data') / 'AMSEAS-subset.nc'
ds = xr.open_dataset(data_path)
print('Dataset loaded:', ds)
```

```python
print('Grid type:', ds.xsg.grid)
print('Grid vars:', ds.xsg.grid_vars)
print('Data vars:', ds.xsg.data_vars)
```

```python
lats = ds.cf['latitude'].values
lons = ds.cf['longitude'].values
lat_mid = (float(lats.min()) + float(lats.max())) / 2
lon_mid = (float(lons.min()) + float(lons.max())) / 2
bbox = (float(lons.min()), float(lats.min()), lon_mid, lat_mid)
ds_sub = ds.xsg.subset_bbox(bbox)
print('Subsetted dataset:', ds_sub)
print('Done.')
```
